```python
## if statement 

age = 20
if age >=18:
    print("you are allowed to vote in the elections")
```

```python
## if statement (take user input) and else 
## The else statement executes a block of code if the condition in the if statement is False.
age1 = int(input("Enter your age"))
if age1>=20:
    print("you are allow to vote in the elections")
else:
    print("you are not allow to vote in the elections")
           
           
```

```python
## elif
## The elif statement allow you to check multiple condtions. it stands for "else if"

age2 = int(input("Enter your age"))
if age2<13:
    print("you are a child")
elif age2<20:
    print("you are a teenager")
else:
    print("you are an adult")
```

```python
## Nested Conditional Statement

# you can place one or more if, elif, or else statement inside another if, elif, or else statement to create conditional statements.

## number even, odd, negative :- this are three condition my condition 

num = int(input("Enter the number"))
if num>0:
    print("The number is positive")
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is zero or negative")

```

```python
## parcical example 

## Determine if  a year is a leap year using condition statement

year = int(input("Enter the Year"))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"is a leap year")
        else:
            print(year,"is a not leap year")
    
    else:
        print(year,"is a leap year")
else:
    print(year,"is a not leap year")
        
```
