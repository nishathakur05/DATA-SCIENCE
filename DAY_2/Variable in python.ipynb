# 1.Variable 
Variables are fundamental elements in programming used to store data that can be referenced and manipulated in a program. In Python, variables are created when you assign a value to them, and they do not need explicit declaration to reserve memory space. The declaration happens automatically when you assign a value to a variable.

# Why do we use variables?
:- We use variables to store values and give them meaningful names.
* for example:Suppose 7321830841 is a phone number. If we see only the number, we do not know whose number it is.

But when we write:

```python
Nisha = 77628374
print(Nisha)
```

NOTE(Variable): Now we understand that this is Nisha’s phone number.

In the same way, in programming languages, variables are used to store different types of data such as numbers, strings, and other values, so that the program becomes easy to understand and use.

```python
age =25 
print(type(age))

# Type conversion 
age_str=str(age)
print(age_str)
print(type(age_str))

```

```python
## Dynamic Typing 
## Python allows the type of variable to change as the program executes 
var=10 # int
print(var,type(var))

var="hello"
print(var,type(var))

var=3.14
print(var,type(var))
```
