# 3.Indentation :- 
Indentation in Python is used to show the structure and levels of the code. Unlike many other programming languages that use braces {} to mark blocks of code, Python uses indentation (spaces at the beginning of a line) to group statements. This means all the statements in the same block must have the same level of indentation.

    * Indentation is used in:

       1.if statements
       2.loops (for, while)
       3.functions
       4.classes
       5.try/except blocks


```python
# for example indentation:- 
age = 32
if age >30:
    print(age)

print(age)
```
